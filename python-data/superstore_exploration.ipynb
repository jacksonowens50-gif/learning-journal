```python
region = df.groupby('Region')[['Sales','Profit']].sum()
region['Margin %'] = (region['Profit'] / region['Sales'] * 100).round(1)
region.sort_values('Margin %')
```

```python
df.groupby('Year')[['Sales','Profit']].sum()
```

```python
yearly = df.groupby('Year')[['Sales','Profit']].sum()
yearly['Sales YoY %'] = (yearly['Sales'].pct_change() * 100).round(1)
yearly
```

```python
df['Ship Mode'].value_counts()
```

```python
df['Ship Mode'].value_counts(normalize=True) * 100
```

```python
pd.crosstab(df['Ship Mode'], df['Segment'], normalize='columns') * 100
```

```python
disc = df.groupby('Sub-Category').agg(
    AvgDiscount=('Discount','mean'),
    TotalProfit=('Profit','sum')
)
disc.sort_values('AvgDiscount', ascending=False)
```

```python
df.groupby('Discount').agg(
    Lines=('Profit','size'),
    AvgProfit=('Profit','mean')
)
```

```python
top_sales  = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
top_profit = df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(10)
```
